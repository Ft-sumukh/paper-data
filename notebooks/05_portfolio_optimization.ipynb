### Optimizing Equal Weight, Global Minimum Variance, Mean-Variance, and Risk Parity

```python
import sys
sys.path.append('..')
import pandas as pd
from src.optimization import get_optimizer
from src.features.returns import calculate_simple_returns, calculate_covariance_matrix

prices = pd.read_csv('../data/processed/prices_clean.csv', index_col=0, parse_dates=True)
rets = calculate_simple_returns(prices)
mu = rets.mean().values * 252.0
cov = calculate_covariance_matrix(rets, annualize=True)
symbols = list(prices.columns)

for strat in ['equal_weight', 'min_variance', 'mean_variance', 'risk_parity']:
    opt = get_optimizer(strat)
    res = opt.optimize(mu, cov, symbols)
    print(f'=== {strat.upper()} ===')
    print(f'Expected Return: {res.expected_return*100:.2f}%, Volatility: {res.volatility*100:.2f}%, Sharpe: {res.sharpe_ratio:.2f}')
```
