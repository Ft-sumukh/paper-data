### Cross-Asset Returns, Sector Groupings, and Empirical Correlation Matrix

```python
import sys
sys.path.append('..')
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.features.returns import calculate_simple_returns, calculate_correlation_matrix

prices = pd.read_csv('../data/processed/prices_clean.csv', index_col=0, parse_dates=True)
rets = calculate_simple_returns(prices)
corr = calculate_correlation_matrix(rets)

plt.figure(figsize=(10, 8))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', vmin=-0.2, vmax=1.0)
plt.title('Asset Pearson Correlation Matrix')
plt.show()
```
